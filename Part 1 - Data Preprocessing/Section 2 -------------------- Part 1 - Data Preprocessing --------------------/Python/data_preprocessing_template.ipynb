# Data Preprocessing Template

**The big picture:** this is a short **starter template**, a skeleton you copy at the beginning of a new ML project. It does only the steps almost every project needs: load the libraries, load the data, and split it into a training set and a test set.

The data (`Data.csv`) has 10 customers with their **Country**, **Age** and **Salary**, and whether they **Purchased** a product (Yes/No). The other cleaning steps (filling missing values, turning words into numbers, scaling) are shown in the *Data Preprocessing Tools* notebook; you add them to this template only when your data needs them.

### Before you start
- **Read first:** **Data Preprocessing Tools** — this template is a cut-down version of it, so it helps to know what the removed steps (filling gaps, encoding, scaling) do.
- **New words in this notebook:** *starter template*, *iloc*, *train_test_split*, *random_state*. Look them up in the [glossary](../../../GLOSSARY.md) if you get stuck.

## Importing the libraries

- **pandas** reads and handles tables, like Excel in Python.
- **numpy** does fast maths on lists of numbers.
- **matplotlib** draws charts. It isn't used yet, but most projects built from this template will need it.

```
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

- Reads `Data.csv` into a table.
- **X** = every column except the last (Country, Age, Salary). These are the **inputs**.
- **y** = the last column, *Purchased*. This is what we want to **predict**.
- `.values` turns the table into a plain grid of values that scikit-learn (the ML library) can use.

**Why "all but the last column"?** It's a convention: put the answer column last in your CSV, and this same code works for any dataset without changes.

```
dataset = pd.read_csv('Data.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

## Splitting the dataset into the Training set and Test set

- `test_size = 0.2` → **80% of the rows (8 customers)** go into the *training set*, which the model learns from.
- **20% of the rows (2 customers)** go into the *test set*, held back to check the model afterwards.
- `random_state = 0` makes the shuffle come out the same every time you run it.

**Why?** Testing on data the model has never seen shows whether it actually learned the pattern rather than memorising the answers.

Note: in this template the data is *not* cleaned. X still contains country names and two missing values, so a model couldn't use it yet.

**In one sentence:** a reusable starting point that loads the libraries and the data, separates the inputs from the answer, and keeps 20% of the rows aside for testing.

```
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

## Common mistakes to avoid
- **Thinking the data is ready for a model after this template.** Here X still contains country names and two `nan` gaps, so most models would crash. Add the cleaning steps from the Tools notebook whenever your data needs them.
- **Using a CSV where the answer isn't the last column.** `iloc[:, :-1]` and `iloc[:, -1]` assume the target is last. If it isn't, X will contain the answer and y will be the wrong column — move the column or change the slices.
- **The CSV not being found.** `pd.read_csv('Data.csv')` looks in the folder the notebook is running from. If you copy the template elsewhere, also copy the CSV or give its full path, or you'll get a `FileNotFoundError`.
- **Changing `random_state` between runs and comparing results.** Each `random_state` gives a different shuffle, so results from different seeds aren't a like-for-like comparison. Keep it fixed while you experiment with other settings.
